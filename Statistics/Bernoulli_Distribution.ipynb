```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data_files/light_bulbs_quality_control.csv')
df.head()
```

```python
outcomes = df['Test_Result']
outcomes
```

```python
outcomes.head()
```

```python
p_mean = outcomes.mean()
p_mean
```

```python
#calculating of pmf
p_0 = (1-p_mean)**(1-0) * p_mean**0
p_1 = (1-p_mean)**(1-1) * p_mean**1

print(p_0, p_1)
```

```python
def bernuili(p_mean, x):
    return (1-p_mean)**(1-x) * p_mean**x

pmf_0 = bernuili(p_mean, 0)
pmf_1 = bernuili(p_mean, 1)
print(pmf_0, pmf_1)
```

```python
#varience of Bernoulli distribution
varience = p_mean * (1-p_mean)
varience
```

```python
#std_dev of Bernoulli distribution
std_dev = np.sqrt(varience)
std_dev
```

```python
#calculate skewness of Bernoulli distribution
skewness = (1 - 2*p_mean) / std_dev
skewness
```
