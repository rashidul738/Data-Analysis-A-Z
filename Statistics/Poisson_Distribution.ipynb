```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.stats import poisson
```

```python
data = pd.read_csv("../data_files/poisson_dataset.csv")
data.head()
```

```python
data["Event_Type"].unique()
```

```python
data.head(2)
```

```python
data["p(Observed_Count)"] = data.apply(lambda row: poisson.pmf(row["Observed_Count"], mu = row["Average_Rate"]), axis = 1)
data.head()
```

```python
data['Observed_Count'].unique().size
```

```python
plt.figure(figsize = (12, 6))
plt.subplot(1, 2, 1)
plt.hist(data['Observed_Count'], bins = 20, edgecolor = 'black', color = '#D473D4', alpha = 0.7)
plt.xlabel('Observed Count')
plt.ylabel('Frequency')
 
plt.title("Poisson Distribution")
# plt.show()

#scatter plot
plt.subplot(1, 2, 2)
plt.scatter(data['Average_Rate'], data['p(Observed_Count)'], color = 'blue', alpha = 0.7, edgecolor = 'red')
plt.xlabel('Average Rate by Lambda')
plt.ylabel('Probabolity of Observed Count')
plt.title("Avarage Rate vs P(Observed Count)")
plt.show()
```
